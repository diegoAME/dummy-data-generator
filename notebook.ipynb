```python
# Ejemplo de uso: generar un registro sintético a partir de un esquema Spark
def generate_sparklike_pd_df(source_engine,schema,number_of_records):
    output_dataset = []

    if source_engine == "spark":
        reference_dict = spark_types

    if source_engine == "postgres":
        reference_dict = postgres_types

    if source_engine == "mysql":
        reference_dict = mysql_types

    if source_engine == "mongodb":
        reference_dict = mongodb_types

    for i in range(number_of_records):
        output_dataset.append({col: reference_dict[tipo]() for col, tipo in schema.items()})

    return pd.DataFrame(output_dataset)

# Ejemplo de esquema Spark
esquema_ejemplo = {
    'id': 'UUID',
    'nombre': 'VARCHAR',
    'activo': 'BOOLEAN',
    'fecha_registro': 'TIMESTAMP',
    'saldo': 'REAL',
    'datos': 'VARCHAR'
}

# Generar un registro de ejemplo
generate_sparklike_pd_df(source_engine="postgres",schema=esquema_ejemplo,number_of_records=100)
#generate_sparklike_pd_df(source_engine="spark",schema=esquema_ejemplo,number_of_records=100).to_csv("faker_sample.csv",index=False)
```
